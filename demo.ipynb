# Reservoir Proxy Model — Interactive Demo

This notebook lets you run inference with all four trained models in one place.

**Before running:** make sure the `saved_models/` folder exists alongside this notebook.
It is created automatically when you run each training notebook to completion.

---
## Step 1 — Set your reservoir scenario
Edit the six parameters in the cell below, then run all cells.

```python
# ── YOUR SCENARIO ────────────────────────────────────────────────────
# Change these values to predict a different reservoir scenario

scenario = {
    'producer_bhp_psi':      3000,   # Producer bottom-hole pressure (psi)
    'gas_inj_rate_mscf_d':     35.0, # Gas injection rate (Mscf/day)
    'inj_bhp_limit_psi':    10000,   # Injector BHP limit (psi)
    'init_prod_period_days':  3230,  # Initial production period (days)
    'perm_multiplier':          1.0, # Permeability multiplier (0.1 – 10)
    'poro_multiplier':          1.0, # Porosity multiplier (0.5 – 2.0)
}
# ─────────────────────────────────────────────────────────────────────

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib, json, warnings
warnings.filterwarnings('ignore')

SAVED    = 'saved_models'
FEATURES = list(scenario.keys())
df_in    = pd.DataFrame([scenario])
print('Scenario loaded:', scenario)
```

## Step 2 — Quick scalar prediction (MLP, PINN, Random Forest)

These three models predict four final/peak values instantly.

```python
import tensorflow as tf

SCALAR_TARGETS = ['final_fopt', 'final_fpr', 'final_fopr', 'peak_fopr']
SCALAR_UNITS   = {'final_fopt':'STB', 'final_fpr':'psi',
                  'final_fopr':'STB/d', 'peak_fopr':'STB/d'}

def scalar_predict(model_path, sx_path, sy_path):
    m  = tf.keras.models.load_model(model_path, compile=False)
    sx = joblib.load(sx_path)
    sy = joblib.load(sy_path)
    return sy.inverse_transform(m.predict(sx.transform(df_in[FEATURES].values), verbose=0))[0]

mlp_pred  = scalar_predict(f'{SAVED}/mlp_reservoir_proxy.keras',
                            f'{SAVED}/mlp_scaler_X.pkl', f'{SAVED}/mlp_scaler_y.pkl')
pinn_pred = scalar_predict(f'{SAVED}/pinn_base_model.keras',
                            f'{SAVED}/pinn_scaler_X.pkl', f'{SAVED}/pinn_scaler_y.pkl')

rf_pred = []
for key in SCALAR_TARGETS:
    rf_pred.append(joblib.load(f'{SAVED}/rf_{key}.pkl').predict(df_in[FEATURES].values)[0])

comparison = pd.DataFrame({
    'MLP':  mlp_pred,
    'PINN': pinn_pred,
    'RF':   rf_pred,
}, index=SCALAR_TARGETS)

comparison.index = [f'{k}  ({SCALAR_UNITS[k]})' for k in SCALAR_TARGETS]
comparison.style.format('{:,.1f}').set_caption('Model comparison — scalar targets')
```

## Step 3 — Full 22-year time-series forecast (Encoder-Decoder LSTM)

The LSTM predicts all 8 physical variables across the full production lifetime.

```python
model_lstm    = tf.keras.models.load_model(f'{SAVED}/enc_dec_lstm_reservoir_proxy.keras', compile=False)
scaler_static = joblib.load(f'{SAVED}/scaler_static.pkl')
scaler_time   = joblib.load(f'{SAVED}/scaler_time.pkl')
scaler_y_lstm = joblib.load(f'{SAVED}/scaler_y.pkl')
avg_time_grid = np.load(f'{SAVED}/lstm_avg_time_grid.npy')

with open(f'{SAVED}/lstm_metadata.json') as f:
    meta = json.load(f)

MAX_LEN        = meta['MAX_LEN']
OUTPUT_TARGETS = meta['OUTPUT_TARGETS']
TARGET_UNITS   = meta['TARGET_UNITS']
ENC_FEATURES   = meta['ENCODER_FEATURES']

X_static = scaler_static.transform(df_in[ENC_FEATURES].values)
X_time   = scaler_time.transform(avg_time_grid.reshape(-1, 1)).reshape(1, MAX_LEN, 1)

y_s   = model_lstm.predict([X_static, X_time], verbose=0)
y_out = scaler_y_lstm.inverse_transform(y_s.reshape(-1, len(OUTPUT_TARGETS)))

df_lstm = pd.DataFrame(y_out, columns=OUTPUT_TARGETS)
df_lstm.insert(0, 'time_days', avg_time_grid)

print('Final values from LSTM:')
for col, unit in zip(OUTPUT_TARGETS, TARGET_UNITS):
    print(f'  {col:<12} {df_lstm[col].iloc[-1]:>12,.2f}  {unit}')
```

## Step 4 — Production curves plot

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
fig.suptitle('LSTM Proxy — 22-Year Production Forecast', fontsize=13, fontweight='bold')

plot_cfg = [
    ('fopr', 'Oil Rate',          'STB/day', '#e74c3c'),
    ('fopt', 'Cumulative Oil',    'STB',     '#2ecc71'),
    ('fpr',  'Reservoir Pressure','PSIA',    '#3498db'),
]

for ax, (col, title, unit, color) in zip(axes, plot_cfg):
    ax.plot(df_lstm['time_days'], df_lstm[col], color=color, linewidth=2.5)
    ax.set_title(title, fontsize=11)
    ax.set_xlabel('Time (days)')
    ax.set_ylabel(unit)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('prediction_plot.png', dpi=150, bbox_inches='tight')
plt.show()
print('Plot saved to prediction_plot.png')
```
